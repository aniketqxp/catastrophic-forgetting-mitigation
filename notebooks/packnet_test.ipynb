# PackNet

## PackNet (Progressive Pruning)

**Key Idea**: Learn task-specific subnetworks by pruning unimportant weights and freezing the critical ones.

**Methodology**: After training a task on available free weights, keep a set percentage (`keep_ratio=0.30`) of the largest-magnitude weights by freezing their gradients. Zero out the rest for use by future tasks. This creates dedicated structural capacity per task, avoiding interference.

### Setup and Data Loading

```
import sys
sys.path.insert(0, '..')

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader
import numpy as np

from src import (
    set_seed, PermutedMNIST, SimpleNN, evaluate_all_tasks,
    calculate_forgetting_metrics, plot_all_results
)

set_seed(42)

num_tasks = 5
input_size = 28 * 28
hidden_size = 256
output_size = 10
learning_rate = 0.01
epochs_per_task = 5
batch_size = 64
keep_ratio = 0.30  # Fraction of FREE weights to keep (freeze) per task

permutations = [torch.randperm(input_size) for _ in range(num_tasks)]
train_tasks = [PermutedMNIST(root="./data", train=True, permutations=permutations[i]) for i in range(num_tasks)]
test_tasks = [PermutedMNIST(root="./data", train=False, permutations=permutations[i]) for i in range(num_tasks)]
train_loaders = [DataLoader(train_tasks[i], batch_size=batch_size, shuffle=True) for i in range(num_tasks)]
test_loaders = [DataLoader(test_tasks[i], batch_size=batch_size, shuffle=False) for i in range(num_tasks)]
```

### Core PackNet Logic

PackNet works by iteratively:
1. **Training** the network on a new task using only the "free" (unfrozen) weights
2. **Pruning** the free weights  keeping only the top-K% most important ones (by magnitude)
3. **Freezing** those important weights permanently so future tasks cannot modify them

The key insight: the pruned weights were *unimportant* for the current task, so even when
future tasks repurpose them, the current task's performance is preserved.

```
class PackNet:
    """
    PackNet continual learning via structured pruning and freezing.

    After each task, the top `keep_ratio` fraction of currently-free weights
    are identified as "important," then permanently frozen. The remaining
    free weights are zeroed out and released for the next task.
    """
    def __init__(self, model, keep_ratio=0.30):
        self.model = model
        self.keep_ratio = keep_ratio
        # frozen_mask: 1.0 = frozen (belongs to a previous task), 0.0 = free
        self.frozen_masks = {}
        # saved_weights: the frozen weight values (restored after each optimizer step)
        self.saved_weights = {}

        # Initialize all weight masks to "free"
        for name, param in model.named_parameters():
            if 'weight' in name:
                self.frozen_masks[name] = torch.zeros_like(param.data)

    def prune_and_freeze(self, task_idx):
        """After training on a task, prune unimportant free weights and freeze the rest."""
        total_frozen = 0
        total_params = 0

        for name, param in self.model.named_parameters():
            if 'weight' in name:
                frozen = self.frozen_masks[name]
                free_mask = (1.0 - frozen)

                # Get absolute magnitudes of free weights only
                free_abs = param.data.abs() * free_mask
                free_values = free_abs[free_mask.bool()]

                if len(free_values) == 0:
                    total_frozen += int(frozen.sum().item())
                    total_params += frozen.numel()
                    continue

                # Find threshold: keep top keep_ratio fraction of free weights
                k = max(1, int(len(free_values) * self.keep_ratio))
                threshold = torch.topk(free_values.flatten(), k).values[-1].item()

                # Task mask: free weights that survive pruning
                task_mask = ((free_abs >= threshold) & free_mask.bool()).float()

                # Zero out pruned free weights (they become available for future tasks)
                param.data *= (frozen + task_mask).clamp(max=1.0)

                # Update frozen mask
                self.frozen_masks[name] = (frozen + task_mask).clamp(max=1.0)

                # Save frozen weight values
                self.saved_weights[name] = param.data.clone()

                total_frozen += int(self.frozen_masks[name].sum().item())
                total_params += self.frozen_masks[name].numel()

        pct = 100.0 * total_frozen / total_params if total_params > 0 else 0
        print(f"\n[PackNet] After Task {task_idx+1}: {pct:.1f}% of weight params frozen, "
              f"{100-pct:.1f}% free for future tasks")

    def on_after_backward(self):
        """Zero gradients for frozen weights (call AFTER loss.backward(), BEFORE optimizer.step())."""
        for name, param in self.model.named_parameters():
            if name in self.frozen_masks and param.grad is not None:
                param.grad.data *= (1.0 - self.frozen_masks[name])

    def on_after_step(self):
        """Restore frozen weights to saved values (call AFTER optimizer.step())."""
        for name, param in self.model.named_parameters():
            if name in self.saved_weights:
                frozen = self.frozen_masks[name]
                saved = self.saved_weights[name]
                param.data = param.data * (1.0 - frozen) + saved * frozen


def train_task_packnet(model, packnet, task_idx, train_loader, criterion, epochs=5, lr=0.01):
    """Train a single task with PackNet gradient masking."""
    # Fresh optimizer each task  essential so frozen params don't carry stale state
    optimizer = optim.SGD(model.parameters(), lr=lr)

    task_train_loss = []
    for epoch in range(epochs):
        model.train()
        running_loss = 0.0
        correct, total = 0, 0

        for inputs, labels in train_loader:
            outputs = model(inputs)
            loss = criterion(outputs, labels)

            optimizer.zero_grad()
            loss.backward()

            # Mask frozen gradients BEFORE step
            packnet.on_after_backward()

            optimizer.step()

            # Restore frozen weights AFTER step
            packnet.on_after_step()

            running_loss += loss.item()
            _, predicted = torch.max(outputs.data, 1)
            total += labels.size(0)
            correct += (predicted == labels).sum().item()

        epoch_loss = running_loss / len(train_loader)
        epoch_acc = 100 * correct / total
        task_train_loss.append(epoch_loss)
        print(f'Task {task_idx+1}, Epoch {epoch+1}/{epochs}, '
              f'Loss: {epoch_loss:.4f}, Accuracy: {epoch_acc:.2f}%')

    # Prune and freeze after training
    packnet.prune_and_freeze(task_idx)

    return task_train_loss
```

### Sequential Training Loop with PackNet

```
model = SimpleNN(input_size, hidden_size, output_size)
packnet = PackNet(model, keep_ratio=keep_ratio)
criterion = nn.CrossEntropyLoss()

training_losses = []
task_accuracies_history = []
initial_accuracies = []

for task_idx in range(num_tasks):
    print(f"\n{'='*50}\nTraining on Task {task_idx+1}\n{'='*50}")

    task_loss = train_task_packnet(
        model, packnet, task_idx, train_loaders[task_idx],
        criterion, epochs=epochs_per_task, lr=learning_rate
    )
    training_losses.extend(task_loss)

    print("\nEvaluating on all tasks seen so far:")
    task_accs = evaluate_all_tasks(model, test_loaders[:task_idx+1])

    task_accuracies_history.append(task_accs)
    initial_accuracies.append(task_accs[0] if task_idx == 0 else task_accs[-1])
```

### Metrics & Visualization

```
forgetting_rate = calculate_forgetting_metrics(task_accuracies_history, initial_accuracies)
plot_all_results(training_losses, task_accuracies_history, forgetting_rate, num_tasks, method_name='PackNet')
```
