# Baseline Class

## Baseline Class-Incremental (No Mitigation)

**Algorithm**: Standard sequential training on split classes without any form of memory or regularization.

**Expected Behavior**: Severe catastrophic forgetting, as the model will optimize weights solely for the newest classes, completely destroying the representations necessary for older classes.

### Setup and Data Loading

```python
import sys
sys.path.insert(0, '..')

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader

from src import (
    set_seed, ClassIncrementalMNIST, SimpleNN, 
    train_task, evaluate_cumulative_tasks,
    calculate_forgetting_metrics,
    plot_learning_curves, plot_task_accuracies, 
    display_accuracy_table, plot_forgetting_rates
)

set_seed(42)

num_tasks = 5
classes_per_task = 2
input_size = 28 * 28
hidden_size = 256
output_size = 10
learning_rate = 0.01
epochs_per_task = 5
batch_size = 64

class_splits = [list(range(i * classes_per_task, (i + 1) * classes_per_task)) for i in range(num_tasks)]

train_tasks = [ClassIncrementalMNIST(root="./data", train=True, classes=class_splits[i]) for i in range(num_tasks)]
test_tasks = [ClassIncrementalMNIST(root="./data", train=False, classes=class_splits[i]) for i in range(num_tasks)]

train_loaders = [DataLoader(train_tasks[i], batch_size=batch_size, shuffle=True) for i in range(num_tasks)]
```

### Sequential Training Loop

```python
model = SimpleNN(input_size, hidden_size, output_size)
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=learning_rate)

training_losses = []
task_accuracies_history = []
initial_accuracies = []

for task_idx in range(num_tasks):
    print(f"\n{'='*50}\nTraining on Task {task_idx+1}: Classes {class_splits[task_idx]}\n{'='*50}")
    
    task_loss, _ = train_task(model, task_idx, train_loaders[task_idx], criterion, optimizer, epochs=epochs_per_task)
    training_losses.extend(task_loss)
    
    print("\nEvaluating on cumulative tasks seen so far:")
    acc = evaluate_cumulative_tasks(model, test_tasks[:task_idx + 1])
    print(f"Cumulative Model Accuracy after Task {task_idx + 1}: {acc:.2f}%")
    
    # Note: cumulative accuracy doesn't provide perfect granular task decay tracking,
    # but we store it for basic visualization
    # For real forgetting metrics, evaluate_all_tasks is better, but this is class-incremental baseline
    task_accuracies_history.append([acc])
    initial_accuracies.append(acc)
```

### Metrics & Visualization

```python
plot_learning_curves(training_losses, title="Baseline Class-Incremental Loss")

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(task_accuracies_history) + 1), [a[0] for a in task_accuracies_history], marker='o', linewidth=2)
plt.xlabel('Task Number')
plt.ylabel('Cumulative Accuracy (%)')
plt.title('Class-Incremental Learning: Baseline Performance')
plt.grid(True, alpha=0.3)
plt.show()
```
