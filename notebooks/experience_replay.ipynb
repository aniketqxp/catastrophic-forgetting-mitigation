# Experience Replay

## Experience Replay

**Key Idea**: Store past task data and replay it systematically.

**Methodology**: Uses a memory buffer managed by **Reservoir Sampling** to maintain a perfectly uniform historical distribution. During training, batches are constrained to a strict 1:1 balance between current task data and historical replay data. This eliminates the recency bias.

### Setup and Data Loading

```python
import sys
sys.path.insert(0, '..')

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader

from src import (
    set_seed, ClassIncrementalMNIST, SimpleNN, ReplayBuffer,
    train_task_with_replay, evaluate_cumulative_tasks,
    plot_learning_curves
)

set_seed(42)

num_tasks = 5
classes_per_task = 2
input_size = 28 * 28
hidden_size = 256
output_size = 10
learning_rate = 0.005
epochs_per_task = 5
batch_size = 32
replay_capacity = 2000

class_splits = [list(range(i * classes_per_task, (i + 1) * classes_per_task)) for i in range(num_tasks)]

train_tasks = [ClassIncrementalMNIST(root="./data", train=True, classes=class_splits[i]) for i in range(num_tasks)]
test_tasks = [ClassIncrementalMNIST(root="./data", train=False, classes=class_splits[i]) for i in range(num_tasks)]

train_loaders = [DataLoader(train_tasks[i], batch_size=batch_size, shuffle=True) for i in range(num_tasks)]
```

### Sequential Training Loop with Experience Replay

```python
model = SimpleNN(input_size, hidden_size, output_size)
criterion = nn.CrossEntropyLoss()
optimizer = optim.SGD(model.parameters(), lr=learning_rate)

buffer = ReplayBuffer(replay_capacity)
training_losses = []
cumulative_accuracies = []

for task_idx in range(num_tasks):
    print(f"\n{'='*50}\nTraining on Task {task_idx+1}: Classes {class_splits[task_idx]}\n{'='*50}")
    
    # Train with experience replay
    task_loss, _ = train_task_with_replay(
        model, task_idx, train_loaders[task_idx], criterion, optimizer, 
        buffer, epochs=epochs_per_task, batch_size=batch_size
    )
    training_losses.extend(task_loss)
    
    # Evaluate on cumulative tasks
    acc = evaluate_cumulative_tasks(model, test_tasks[:task_idx + 1])
    cumulative_accuracies.append(acc)
    print(f"Cumulative Model Accuracy after Task {task_idx + 1}: {acc:.2f}%")
    
    # Add current task data to buffer for future replay
    current_task_data = torch.stack([x for x, _ in train_tasks[task_idx]])
    current_task_labels = [y for _, y in train_tasks[task_idx]]
    buffer.add(current_task_data, current_task_labels)
```

### Metrics & Visualization

```python
plot_learning_curves(training_losses, title="Experience Replay Loss")

import matplotlib.pyplot as plt
plt.figure(figsize=(10, 5))
plt.plot(range(1, len(cumulative_accuracies) + 1), cumulative_accuracies, marker='o', linewidth=2)
plt.xlabel('Task Number')
plt.ylabel('Cumulative Accuracy (%)')
plt.title('Class-Incremental Learning: Experience Replay Performance')
plt.grid(True, alpha=0.3)
plt.show()
```
